```python
scores=[45,88,92,67,88,100,73]
scores.sort(reverse=True)
print("After sorting descending:", scores)
scores.remove(88)
print(scores)
scores.insert(0,50)
print(scores)

total=0
maximum=scores[0]
minimum=scores[0]

for num in scores:
    total+=num
    if num > maximum:
        maximum=num
    if num < minimum:
        minimum=num

print("Sum:",total)
print("Max:",maximum)
print("Min:",minimum)
```

```python
def min_max(nums):
    minimum=nums[0]
    maximum=nums[0]
    for num in nums:
        if num < minimum:
            minimum=num
        if num > maximum:
            maximum=num
    return(minimum,maximum)
    
numbers=[45,88,92,67,88,100,73]
low,high=min_max(numbers)
print("Min:",low)
print("Max:",high)
```

```python
text="the quick brown fox jumps over the lazy dog the fox is quick"
words=text.split()
print(words)

word_counts={}
for word in words:
    word_counts[word]=word_counts.get(word,0)+1
print(word_counts)

top_3=sorted(word_counts.items(), key=lambda x:x[1], reverse=True)[:3]
print(top_3)

unique_words=set(words)
print(unique_words)
```

```python
grades={
    "Alice":[88,92,79],
    "Bob":[70,65,90],
    "Carol":[95,98,100]   
}
averages={}
for student, marks in grades.items():
    averages[student]=sum(marks)/len(marks)
print(averages) 

top_student=max(averages, key=averages.get)
print(top_student,":",averages[top_student])

all_grades=set()
for marks in grades.values():
    all_grades.update(marks)
print("Unique grades:",all_grades)
```

```python
cart_list=[("apple",1.50),("bread",3.00),("milk",2.50)]
cart=dict(cart_list)
print(cart)
cart.update({"eggs":4.00,"cheese":5.50})
print(cart)
sorted_cart=sorted(cart.items(), key=lambda x:x[1])
print(sorted_cart)
most_expensive=max(cart, key=cart.get)
removed_price=cart.pop(most_expensive)
print(f"Removed {most_expensive} (${removed_price})")
print(cart)
```
