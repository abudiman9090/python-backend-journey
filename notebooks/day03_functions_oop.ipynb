```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"
print(greet("Ali"))
print(greet("Ali","Hi"))
```

```python
def square(n):
    return n*n
print(square(5))
x=(square(5))
print(x+1)
```

```python
def average(*nums):
    return sum(nums)/len(nums)
average(1,2,3)
```

```python
def describe(**info):
    for key, value in info.items():
        print(f"{key:}, {value}")
describe(name="Ali", age=22)
```

```python
class Book:
    def __init__(self,title,author):
        self.title=title
        self.author=author
        
    def __str__(self):
        return f"{self.title} by {self.author}"
        
b1=Book("The Hobbit", "Tolkien")
b2=Book("1984", "Orwell")
print(b1)
print(b2)

class Ebook(Book):
    def __init__(self, title, author, size_mb):
        super().__init__(title, author)
        self.size_mb=size_mb

    def __str__(self):
        return f"{super().__str__()} ({self.size_mb}mb)"
        
e=Ebook("Sapiens", "Harari", 12)
print(e)
```
