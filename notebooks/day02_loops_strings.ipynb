```python
for n in range(1,51):
    if n%15==0:
        print("FizzBuzz")
    elif n%3==0:
        print("Fizz")
    elif n%5==0:
        print("Buzz")
    else:
        print(n)
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(f"{i} x {j} = {i*j}")
```

```python
squares_even=[x**2 for x in range(1,21) if x%2==0]
print(squares_even)

words=["apple","cat","banana","dog","elephant"]
long_words=[word for word in words if len(word)>4]
print(long_words)

word_lengths={w: len(w) for w in words}
print(word_lengths)
```

```python
text="Hello, World! Programming is fun."
count=0
for ch in text:
    if ch.lower() in "aeiou":
        count += 1
print(count)
```

```python
def is_palindrome(text):
    cleaned=text.replace(" ","").lower()
    return cleaned==cleaned[::-1]
print(is_palindrome("A man a plan a canal Panama"))
print(is_palindrome("Hello World"))
```

```python
nested=[[1,2], [3,4], [5,6]]
flattened = [item for sublist in nested for item in sublist]
print(flattened)

flat=[]
for sublist in nested:
    for item in sublist:
        flat.append(item)
print(flat)
```

```python
sentence= "The quick brown fox jumps over the lazy dog"
word_len_dict={word: len(word) for word in sentence.split()}
print(word_len_dict)
```

```python
sentence= "Hello World Python is great"
reversed_words = ' '.join(word[::-1] for word in sentence.split())
print(reversed_words)
```

```python
for i in range(1,31):
    if i%28==0:
        print("FourSeven")
    elif i%4==0:
        print("Four")
    elif i%7==0:
        print("Seven")
    else:
        print(i)
```
